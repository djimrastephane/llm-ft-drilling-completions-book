Interactive companion to `code/chapter_06/data_quality_gate.py`. Runs Chapter 2's field extractor across the full 76-report archive and flags duplicate field values that deserve a human look before training.

```python
import sys
sys.path.insert(0, "../code/chapter_02")
sys.path.insert(0, "../code/chapter_06")

from data_quality_gate import build_archive_records, find_duplicate_values, classify_duplicate, run_quality_gate

records = build_archive_records()
print(f"{len(records)} reports in the full archive")
```

See exactly which report(s) failed extraction, and why -- open the file yourself to compare its layout against `code/chapter_02/build_training_examples.py`'s field patterns.

```python
failed = [r for r in records if r["status"] != "ok"]
for r in failed:
    print(r["file"])
```

Try a different field -- or lower the bar to see near-duplicates you'd want a fuzzier check for.

```python
for value, nums in find_duplicate_values(records, "activity_planned").items():
    print(f"{classify_duplicate(nums):14} reports {nums}: {value!r}")
```
