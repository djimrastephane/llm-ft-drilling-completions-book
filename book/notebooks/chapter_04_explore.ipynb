# Chapter 4 explore: Tokenization and Embeddings for Domain Fine-Tuning

Interactive companion to `code/chapter_04/tokenize_and_embed.py`. Looks at how the base model's tokenizer splits oilfield shorthand, and whether either the base model's raw embeddings or a general-purpose sentence-embedding model recognize that shorthand as equivalent to its own spelled-out meaning.

```python
import sys
sys.path.insert(0, "../code/chapter_01")
sys.path.insert(0, "../code/chapter_04")

from load_local_model import MODEL_NAME, load_model_and_tokenizer
from tokenize_and_embed import tokenize_with_pieces, raw_embedding_similarities, sentence_embedding_similarities

model, tokenizer = load_model_and_tokenizer(MODEL_NAME)
```

Try your own term -- any word or oilfield abbreviation -- and see exactly how the tokenizer splits it.

```python
term = "NPT"  # try other terms from the sample reports: MD, TVD, BOP, WOB...
print(tokenize_with_pieces(tokenizer, term))
```

Compare raw input-embedding similarity against a general-purpose sentence-embedding model, for your own pair of terms.

```python
pairs = [("NPT", "non-productive time"), ("NPT", "national pipe thread")]

print("Raw input-embedding similarity:")
for r in raw_embedding_similarities(model, tokenizer, pairs):
    print(f"  {r['term_a']!r} vs {r['term_b']!r} -> {r['cosine_similarity']:.3f}")

print("\nSentence-embedding similarity:")
for r in sentence_embedding_similarities(pairs):
    print(f"  {r['term_a']!r} vs {r['term_b']!r} -> {r['cosine_similarity']:.3f}")
```
