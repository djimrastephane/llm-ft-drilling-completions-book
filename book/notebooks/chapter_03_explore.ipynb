# Chapter 3 explore: Baseline Prompting: What the Model Gets Wrong Before Fine-Tuning

Interactive companion to `code/chapter_03/baseline_prompting.py`. Runs the unmodified base model against Chapter 2's training examples, with each example's `output` withheld, and records what it says.

```python
import sys, json
sys.path.insert(0, "../code/chapter_01")
sys.path.insert(0, "../code/chapter_02")
sys.path.insert(0, "../code/chapter_03")

from load_local_model import MODEL_NAME, load_model_and_tokenizer
from build_training_examples import build_training_examples
from baseline_prompting import ask_baseline, run_baseline

model, tokenizer = load_model_and_tokenizer(MODEL_NAME)
examples = build_training_examples()
print(f"{len(examples)} baseline questions built from Chapter 2's training examples")
```

Ask the base model a single question, with the real answer withheld, and see both the model's answer and the report's actual value side by side.

```python
result = ask_baseline(model, tokenizer, examples[0])
print(json.dumps(result, indent=2))
```

Run the full 18-question baseline (this takes a couple of minutes on CPU) and see the overall match rate, plus a few individual answers.

```python
results = run_baseline(model, tokenizer, examples)
matched = sum(r["matched_expected"] for r in results)
print(f"Baseline: {matched}/{len(results)} answers contained the report's actual value verbatim\n")

for r in results[:3]:
    print(f"Q: {r['instruction']} ({r['input']})")
    print(f"Expected: {r['expected']}")
    print(f"Model said: {r['base_model_answer']}\n")
```
